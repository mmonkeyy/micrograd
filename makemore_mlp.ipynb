# Context-window MLP — in progress

This notebook currently constructs three-character contexts and demonstrates an embedding lookup. The forward pass, training, evaluation and sampling remain to be implemented. It follows Andrej Karpathy's makemore tutorial.

```python
import torch
import torch.nn.functional as F
import matplotlib.pyplot as plt

%matplotlib inline
```

```python
words = open("names.txt", "r").read().splitlines()
```

```python
chars = sorted(list(set("".join(words))))
stoi = {s: i + 1 for i, s in enumerate(chars)}
stoi["."] = 0
itos = {i: s for s, i in stoi.items()}
print(itos)
```

```python
block_size = 3
X, Y = [], []
for w in words[:5]:
    print(w)
    context = [0] * block_size
    for ch in w + ".":
        ix = stoi[ch]
        X.append(context)
        Y.append(ix)
        print("".join(itos[i] for i in context), "----->", itos[ix])
        context = context[1:] + [ix]  # crop and append

X = torch.tensor(X)
Y = torch.tensor(Y)
```
